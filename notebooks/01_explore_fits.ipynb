```python
from astropy.io import fits
```

```python
path = "../data/barnard_dss.fits"

with fits.open(path) as hdul:
    hdul.info()
```

```python
with fits.open(path) as hdul:
    header = hdul[0].header
    print(header["DATE-OBS"])
```

```python
import matplotlib.pyplot as plt

with fits.open(path) as hdul:
    image = hdul[0].data

plt.figure()
plt.imshow(image, cmap="gray", origin="lower")
plt.colorbar()
plt.title("Barnard's Star, DSS 1991")
plt.show()
```

```python
import numpy as np

low = np.percentile(image, 1)
high = np.percentile(image, 99)
print(low, high)

plt.figure()
plt.imshow(image, cmap="gray", origin="lower", vmin=low, vmax=high)
plt.colorbar()
plt.title("Barnard's Star, DSS 1991")
plt.plot(294, 294, "r+", markersize=15)
plt.plot(288, 209, "c+", markersize=15)
plt.show()
```

```python
from astropy.wcs import WCS
from astropy.nddata import Cutout2D
from astropy.coordinates import SkyCoord
import astropy.units as u

with fits.open(path) as hdul:
    header = hdul[0].header

wcs = WCS(header)

center = SkyCoord("17h57m48.5s", "+04d41m36s", frame="icrs")
size = 5 * u.arcmin

cutout = Cutout2D(image, position=center, size=size, wcs=wcs)
print(cutout.data.shape)
```

```python
plt.figure()
plt.imshow(cutout.data, cmap="gray", origin="lower", vmin=low, vmax=high)
plt.colorbar()
plt.title("Barnard's Star, DSS 1991")
plt.show()
```

```python
from skyblink.coords import hms_to_degrees, dec_to_degrees
from skyblink.cutout import make_cutout

ra = hms_to_degrees(17, 57, 48.5)
dec = dec_to_degrees("+04", 41, 36)

cutout = make_cutout(image, header, ra, dec, 2)
print(cutout.data.shape)
```
